**Business Problem Overview**

The banking sector handles millions of financial transactions every day, making it a prime target for fraudulent activities. With the increasing use of digital payments, detecting fraudulent transactions has become more challenging and critical.

Financial fraud can lead to significant monetary losses and damage customer trust. Therefore, it is essential for organizations to identify suspicious patterns and prevent fraud before it occurs.

In this project, we perform Exploratory Data Analysis (EDA) on banking transaction data from Paisabazaar to identify patterns and factors associated with fraudulent activities. The insights derived will help in improving fraud detection systems and enhancing transaction security.

### Variables Description

- **ID** : Unique identifier for each record

- **Customer_ID** : Unique ID assigned to each customer

- **Month** : Month of the transaction or record

- **Name** : Name of the customer

- **Age** : Age of the customer

- **SSN** : Social Security Number (unique identity of customer)

- **Occupation** : Customer ka profession (job/business)

- **Annual_Income** : Customer ki yearly income

- **Monthly_Inhand_Salary** : Monthly salary after deductions

- **Num_Bank_Accounts** : Number of bank accounts held by the customer

- **Num_Credit_Card** : Number of credit cards owned

- **Interest_Rate** : Interest rate applied on loans/credit

- **Num_of_Loan** : Total number of loans taken by the customer

- **Type_of_Loan** : Loan ka type (personal, home, car, etc.)

- **Delay_from_due_date** : Payment due date se kitne din late hua

- **Num_of_Delayed_Payment** : Total delayed payments

- **Changed_Credit_Limit** : Credit limit me change (increase/decrease)

- **Num_Credit_Inquiries** : Credit check kitni baar ki gayi

- **Credit_Mix** : Type of credit (good, bad, standard)

- **Outstanding_Debt** : Total pending debt amount

- **Credit_Utilization_Ratio** : Credit usage ratio (kitna limit use hua)

- **Credit_History_Age** : Kitne time se credit history exist karti hai

- **Payment_of_Min_Amount** : Minimum payment kiya gaya ya nahi (Yes/No)

- **Total_EMI_per_month** : Har month ka total EMI amount

- **Amount_invested_monthly** : Monthly investment amount

- **Payment_Behaviour** : Payment karne ka pattern (regular/irregular)

- **Monthly_Balance** : Month ke end me remaining balance

- **Credit_Score** : Customer ka credit score (Target Variable)

##- 1 Number of Customer By Occupation - Bar Char(Demographic Analysis)

```python
occupation_counts=df['Occupation'].value_counts().reset_index()
print(occupation_counts)
```

```python
#baar chart in occupation_counts
df['Occupation'].value_counts().plot(kind='bar',figsize=(15,4))
plt.title("Occupation-wise Number of Customers")
plt.xlabel("Occupation")
plt.ylabel("No of customers")
plt.show()

```

## Why this chart:
We used a bar chart because occupation is a categorical variable, and bar charts are best for comparing values across different categories. Other charts like line charts are not suitable because there is no continuous trend.

#Insights:
- Lawyer occupation has the highest number of customers.
- Writer and Musician have the lowest customer count.
- Customer distribution is almost balanced across occupations.
#Actions:
- Focus more on occupations with higher customers.
- Monitor occupation-wise fraud patterns.
- Improve security checks for high-customer occupations.


```python
sns.histplot(df['Age'],color='blue')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

##Why this chart:
We used a histogram because age is a continuous numerical variable, and histograms help show how data is distributed across ranges. Bar charts are less effective for showing continuous data distribution.

## Chart -3 Annual income by Occupation - Boxplot


```python
occupation_wise_income=df.groupby('Occupation')['Annual_Income'].sum().reset_index()
print(occupation_wise_income)
```

```python
#create  a box plot for Annual Income across Occupation
plt.figure(figsize=(10,5))
sns.boxplot(x='Occupation',y='Annual_Income',data=df)
plt.title('Annual Income by Occupation')
plt.xlabel("Occupation")
plt.ylabel("Annual Income")
plt.xticks(rotation=90)
plt.show()
```

##Why this chart:
We used a box plot because it helps compare income distribution, median, spread, and outliers across occupations. A bar chart would only show average values and hide income variation.

#Insights:

- Annual income is almost similar across different occupations.

- Some customers have very high income (outliers).

- Most customers have income in the middle range.

#Actions:

- Focus more on high-income customers.

- Monitor unusual high-income transactions for fraud detection.

- Create special banking offers for different income groups.
